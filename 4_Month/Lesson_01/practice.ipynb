## 🧠 5. OpenRouter pratformasida bor modellar

| Model | Vazifasi |
|---|---|

| `sensenova-u1.5-lite` | Yengil AI vazifalari va matn bilan ishlash |

```python
from openai import OpenAI
from getpass import getpass

api_key = getpass("SenseNova API Key: ").strip()

client = OpenAI(
    api_key=api_key,
    base_url="https://token.sensenova.ai/v1"
)

response = client.chat.completions.create(
    model="sensenova-6.8-flash-lite",
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant. Always answer in English."
        },
        {
            "role": "user",
            "content": "What is Artificial Intelligence?"
        }
    ],
    max_tokens=1000
)

message = response.choices[0].message

if message.content:
    print("SenseNova Response:")
    print(message.content)

elif message.reasoning:
    print("Reasoning:")
    print(message.reasoning)

else:
    print("No response generated.")

print("\nFinish reason:", response.choices[0].finish_reason)
```
