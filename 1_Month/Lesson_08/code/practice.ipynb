```python
from transformers import AutoModel, AutoTokenizer
from huggingface_hub import hf_hub_download
```

```python
model = AutoModel.from_pretrained("gpt2")
tokenizer = AutoTokenizer.from_pretrained("gpt2")

# Download tokenizer files directly
files = [
    "tokenizer.json",
    "tokenizer_config.json",
    "vocab.json",
    "merges.txt"
]

for file in files:
    path = hf_hub_download(
        repo_id="openai-community/gpt2",
        filename=file,
        local_dir="./gpt2"
    )
    print(path)
```

```python
print(model)
```

```python
# Check embedding layers
print(model.wte)
print(model.wpe)
```

```python
print(model.wte.weight.shape)
```

```python
print(model.wte.weight[40])
```

```python
text = "My name is Ali. I live in Korea"

ids = tokenizer.encode(text)

print("Token IDs:", ids)
print("Tokens:", tokenizer.convert_ids_to_tokens(ids))
```

```python
# Convert tokens to embeddings
inputs = tokenizer(text, return_tensors="pt")

embeddings = model.wte(inputs["input_ids"])

print("Embedding shape:", embeddings.shape)
```

```python
# First token's first 10 embedding values
print("First token embedding:", embeddings[0][0][:10])
```

```python
# Test different words
words = ["lion", "wolf", "fox"]

for word in words:
    ids = tokenizer(word, return_tensors="pt")["input_ids"]
    emb = model.wte(ids)

    print(word, emb.shape)
```

```python
model = AutoModel.from_pretrained("Qwen/Qwen2.5-0.5B")
tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen2.5-0.5B")

# Download Qwen tokenizer files directly
files = [
    "tokenizer.json",
    "tokenizer_config.json",
    "vocab.json",
    "merges.txt"
]

for file in files:
    path = hf_hub_download(
        repo_id="Qwen/Qwen2.5-0.5B",
        filename=file,
        local_dir="./qwen"
    )
    print(path)
```

```python
print(model)
```

```python
# Get embedding layer
embedding_layer = model.get_input_embeddings()

print(embedding_layer)

# Vocabulary size and embedding dimensions
print(embedding_layer.weight.shape)

# One embedding vector
print(embedding_layer.weight[40])
```

```python
# Tokenization
text = "My name is Ali. I live in Korea."

ids = tokenizer.encode(text)

print("Token IDs:", ids)
print("Tokens:", tokenizer.convert_ids_to_tokens(ids))
```

```python
# Convert tokens to embeddings
inputs = tokenizer(text, return_tensors="pt")

embeddings = embedding_layer(inputs["input_ids"])

print("Embedding shape:", embeddings.shape)
```

```python
# First token's first 10 embedding values
print("First token embedding:", embeddings[0][0][:10])
```

```python
# Test different words
words = ["lion", "wolf", "fox"]

for word in words:
    ids = tokenizer(word, return_tensors="pt")["input_ids"]
    emb = embedding_layer(ids)

    print(word, emb.shape)
```
