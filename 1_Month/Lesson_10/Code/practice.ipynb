```python
from transformers import AutoTokenizer, AutoModelForCausalLM
from huggingface_hub import hf_hub_download
```

```python
tokenizer = AutoTokenizer.from_pretrained("facebook/opt-125m")
model = AutoModelForCausalLM.from_pretrained("facebook/opt-125m")
```

```python
# Download tokenizer files directly
files = [
    "tokenizer_config.json",
    "vocab.json",
    "merges.txt"
]

for file in files:
    path = hf_hub_download(
        repo_id="facebook/opt-125m",
        filename=file,
        local_dir="./opt125m"
    )
    print(path)
```

# Tokenization -> Embeddings

```python
text = "My name is Ali. I learn LLM"

ids = tokenizer.encode(text)

print("Token IDs:", ids)
print("Tokens:", tokenizer.convert_ids_to_tokens(ids))
```

```python
# Convert to Embeddings
inputs = tokenizer(text, return_tensors="pt")

embeddings = model.model.decoder.embed_tokens(inputs["input_ids"])

print("Embedding shape:", embeddings.shape)
```

```python
print(embeddings[0][0][:10])
```

```python
words = ["winter", "summer", "spring"]

for word in words:
    ids = tokenizer(word, return_tensors="pt")["input_ids"]

    emb = model.model.decoder.embed_tokens(ids)

    print(word, emb.shape)
```
